```python
import pandas as pd
import numpy as np
import torch
from torch.utils.data import TensorDataset, DataLoader
from transformers import TimeSeriesTransformerConfig, TimeSeriesTransformerForPrediction

# Load and preprocess data
aapl = pd.read_csv("AAPL.csv")
aapl = aapl.rename(columns={"Close/Last": "Close"})
aapl.columns = ["Date", "Close", "Volume", "Open", "High", "Low"]
price_cols = ["Close", "Open", "High", "Low"]

for col in price_cols:
    aapl[col] = aapl[col].replace('[\$,]', '', regex=True).astype(float)

aapl['Date'] = pd.to_datetime(aapl['Date'])
aapl = aapl.sort_values('Date').reset_index(drop=True)

# Create time features
def create_time_features(df):
    df['month'] = df['Date'].dt.month
    df['day_of_week'] = df['Date'].dt.dayofweek
    df['day_of_month'] = df['Date'].dt.day
    df['day_of_year'] = df['Date'].dt.dayofyear
    df['year'] = df['Date'].dt.year
    df['year_sin'] = np.sin(2 * np.pi * df['day_of_year'] / 365.25)
    df['year_cos'] = np.cos(2 * np.pi * df['day_of_year'] / 365.25)
    df['week_sin'] = np.sin(2 * np.pi * df['day_of_week'] / 7)
    df['week_cos'] = np.cos(2 * np.pi * df['day_of_week'] / 7)
    df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
    df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)
    return df

aapl = create_time_features(aapl)
time_features = ['year_sin', 'year_cos', 'week_sin', 'week_cos', 
                'month_sin', 'month_cos', 'day_of_week', 
                'day_of_month', 'month', 'day_of_year', 'year']

# Model parameters
context_length = 30
prediction_length = 5
max_lag = 30  # Matches largest lag in lags_sequence
input_length = context_length + max_lag

# Prepare data windows
ohlc = aapl[['Open', 'High', 'Low', 'Close']].values
time_feats = aapl[time_features].values
n = len(ohlc)
num_windows = n - (input_length + prediction_length) + 1

# Initialize arrays
past_values_arr = np.zeros((num_windows, input_length, 4))
past_time_features_arr = np.zeros((num_windows, input_length, len(time_features)))
future_values_arr = np.zeros((num_windows, prediction_length, 4))
future_time_features_arr = np.zeros((num_windows, prediction_length, len(time_features)))

# Create windows
for i in range(num_windows):
    past_start = i
    past_end = i + input_length
    future_start = past_end
    future_end = past_end + prediction_length
    
    if future_end > len(ohlc):
        break
    
    past_values_arr[i] = ohlc[past_start:past_end]
    past_time_features_arr[i] = time_feats[past_start:past_end]
    future_values_arr[i] = ohlc[future_start:future_end]
    future_time_features_arr[i] = time_feats[future_start:future_end]

# Trim to actual windows created
num_actual_windows = i
past_values_arr = past_values_arr[:num_actual_windows]
past_time_features_arr = past_time_features_arr[:num_actual_windows]
future_values_arr = future_values_arr[:num_actual_windows]
future_time_features_arr = future_time_features_arr[:num_actual_windows]

# Convert to tensors
past_values = torch.tensor(past_values_arr, dtype=torch.float32)
past_time_features = torch.tensor(past_time_features_arr, dtype=torch.float32)
future_values = torch.tensor(future_values_arr, dtype=torch.float32)
future_time_features = torch.tensor(future_time_features_arr, dtype=torch.float32)

# Create mask with full input shape [batch, seq_len, num_features]
# Modify the mask creation to use context_length for scaling
past_observed_mask = torch.ones(
    (past_values.shape[0], context_length, past_values.shape[2]),  # Use context_length here
    dtype=torch.bool
)

# Normalization
means = past_values.mean(dim=(0, 1), keepdim=True)
stds = past_values.std(dim=(0, 1), keepdim=True)
past_values = (past_values - means) / stds
future_values = (future_values - means) / stds

# Train/validation split
train_size = int(0.8 * num_actual_windows)
train_dataset = TensorDataset(
    past_values[:train_size], 
    past_time_features[:train_size],
    past_observed_mask[:train_size],
    future_values[:train_size],
    future_time_features[:train_size]
)

val_dataset = TensorDataset(
    past_values[train_size:], 
    past_time_features[train_size:],
    past_observed_mask[train_size:],
    future_values[train_size:],
    future_time_features[train_size:]
)

batch_size = 32
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)

# Model configuration
config = TimeSeriesTransformerConfig(
    prediction_length=prediction_length,
    context_length=context_length,
    input_size=4,
    num_time_features=len(time_features),
    lags_sequence=[1, 7, 30],
    scaling="std",
    num_static_categorical_features=0,
    num_static_real_features=0,
    num_parallel_samples=100,
    encoder_layers=4,
    decoder_layers=4,
    d_model=64,
)


model = TimeSeriesTransformerForPrediction(config)

# Training loop
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3)
num_epochs = 10

for epoch in range(num_epochs):
    model.train()
    train_loss = 0.0
    for batch in train_loader:
        past_v, past_t, past_m, future_v, future_t = batch
        optimizer.zero_grad()
        outputs = model(
            past_values=past_v,
            past_time_features=past_t,
            past_observed_mask=past_m,
            future_values=future_v,
            future_time_features=future_t
        )
        loss = outputs.loss
        loss.backward()
        optimizer.step()
        train_loss += loss.item()
    
    avg_train_loss = train_loss / len(train_loader)
    val_loss = 0.0
    model.eval()
    with torch.no_grad():
        for batch in val_loader:
            past_v, past_t, past_m, future_v, future_t = batch
            outputs = model(
                past_values=past_v,
                past_time_features=past_t,
                past_observed_mask=past_m,
                future_values=future_v,
                future_time_features=future_t
            )
            val_loss += outputs.loss.item()
    
    avg_val_loss = val_loss / len(val_loader)
    print(f"Epoch {epoch+1}: Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}")

# Prediction generation
data_means = means.squeeze().numpy()
data_stds = stds.squeeze().numpy()

# Prepare final input sequence
last_past_values = ohlc[-input_length:]
last_past_time = time_feats[-input_length:]

# Normalize using training stats
last_past_values_normalized = (torch.tensor(last_past_values, dtype=torch.float32) - means) / stds

# Create future time features
future_dates = pd.date_range(start=aapl['Date'].iloc[-1] + pd.Timedelta(days=1), periods=prediction_length)
future_df = pd.DataFrame({'Date': future_dates})
future_df = create_time_features(future_df)
future_time = torch.tensor(future_df[time_features].values, dtype=torch.float32).unsqueeze(0)

# Generate predictions
model.eval()
with torch.no_grad():
    generated_output = model.generate(
        past_values=last_past_values_normalized.unsqueeze(0),
        past_time_features=torch.tensor(last_past_time, dtype=torch.float32).unsqueeze(0),
        past_observed_mask=torch.ones((1, context_length, 4), dtype=torch.bool),
        future_time_features=future_time,
    )

# Denormalize and format predictions
predicted_prices = generated_output.sequences.mean(dim=1).squeeze().numpy()
predicted_prices = predicted_prices * data_stds + data_means

print("\nPredicted Prices for next 5 days:")
print(predicted_prices)
```
